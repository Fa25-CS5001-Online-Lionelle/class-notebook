Can you go over edge cases? Some unusual inputs are still accounted for without having to write extra code, so by definition would that be an edge case?
* edge case
* null / none case
* invalid 

```python
def check_range(num: int, start: int, end: int) -> bool:
    """ Document it 
    
    inclusive of start, and exclusive of end

    >>> check_range(10, 0, 10)
    False
    >>> check_range(9, 0, 10)
    True
    >>> check_range(0, 0, 10)
    True
    >>> check_range(-1, 0, 10)
    False
    >>> check_range(1, 0, 10)
    True
    """
    if(num < end and num >= start): 
        return True 
    else:
        return False
```

```python
check_range(10, 0, 10)
```

```python
check_range(9, 0, 10)
```

```python
print(check_range(0, 0, 10))
print(check_range(-1, 0, 10))
print(check_range(1, 0, 10))
```

Common but not really useful

```python
print(check_range(5, 0, 10))
print(check_range(-100, 0, 10))
print(check_range(9999, 0, 10000))
```

```python
def count_letters_greater_than(word: str, char: str, num: int) -> int:
    """ 
    >>> count_letters_greater_than("hello", 'l', 1)
    True
    >>> count_letters_greater_than("hello", 'e', 1)
    False
    >>> count_letters_greater_than(None, 'e', 1)
    False
    """
    #if(word is None): return False
    if word and (word.count(char) > num):    
        return True 
    else:
        return False
```

```python
count_letters_greater_than("hello", 'l', 1)
```

```python
count_letters_greater_than("hello", 'e', 1)
```

```python
count_letters_greater_than(None, 'e', 1)
```
